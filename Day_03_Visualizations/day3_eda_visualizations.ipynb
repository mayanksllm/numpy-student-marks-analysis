# Assignment 3: Exploratory Data Analysis & Visualizations
**Dataset:** `retail_sales_visualization_dataset.csv`  
**Libraries:** Pandas, Matplotlib, Seaborn

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')

df = pd.read_csv('retail_sales_visualization_dataset.csv')
df.columns = df.columns.str.strip()
df['Date'] = pd.to_datetime(df['Date'], errors='coerce')
df.head()
```

```python
# 1. Distribution Chart
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.histplot(data=df, x='Sales', bins=25, kde=True, color='#2b5c8f', ax=ax)
ax.set_title('Distribution of Sales per Transaction', weight='bold')
ax.set_xlabel('Sales ($)')
ax.set_ylabel('Transaction Frequency')
plt.tight_layout()
plt.show()
```

**Interpretation 1:** The sales distribution is strongly right-skewed. The vast majority of sales cluster under $5,000, while a few high-value outliers (up to $95,000+) create a long right tail.

```python
# 2. Comparison Chart
cat_sales = df.groupby('Category')['Sales'].sum().reset_index().sort_values(by='Sales')
fig, ax = plt.subplots(figsize=(8, 4.5))
sns.barplot(data=cat_sales, x='Category', y='Sales', palette='viridis', ax=ax)
ax.set_title('Cumulative Sales by Product Category', weight='bold')
ax.set_xlabel('Product Category')
ax.set_ylabel('Total Sales ($)')
plt.tight_layout()
plt.show()
```

**Interpretation 2:** Electronics and Furniture produce the highest cumulative sales volume due to significantly higher unit prices compared to Accessories and Office Supplies.

```python
# 3. Relationship Chart
fig, ax = plt.subplots(figsize=(8.5, 5))
sns.scatterplot(data=df, x='Sales', y='Profit', hue='Category', size='Discount', sizes=(30, 160), alpha=0.85, ax=ax)
ax.set_title('Bivariate Relationship: Sales vs. Profit', weight='bold')
ax.set_xlabel('Sales ($)')
ax.set_ylabel('Profit ($)')
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

**Interpretation 3:** Sales and profit display a robust positive linear relationship. However, transactions with high discounts (size) show reduced profit margins per dollar of sales.

```python
# 4. Correlation Heatmap
fig, ax = plt.subplots(figsize=(7, 5))
corr_matrix = df[['Quantity', 'Unit_Price', 'Discount', 'Sales', 'Profit']].corr()
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='vlag', center=0, ax=ax)
ax.set_title('Correlation Heatmap', weight='bold')
plt.tight_layout()
plt.show()
```

**Interpretation 4:** Sales correlates strongly with Profit (r = 0.88) and Unit Price (r = 0.84). Discounting displays a negative correlation with profit (r = -0.15) without increasing quantity sold (r = -0.13).

```python
# 5. Additional Chart & PNG Export
time_trend = df.set_index('Date').resample('W')['Sales'].sum().reset_index()
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(time_trend['Date'], time_trend['Sales'], marker='o', color='#1b9e77')
ax.set_title('Weekly Aggregate Sales Trend', weight='bold')
ax.set_xlabel('Timeline')
ax.set_ylabel('Sales ($)')
plt.tight_layout()
fig.savefig('weekly_sales_trend.png', dpi=300, bbox_inches='tight')
plt.show()
```

**Interpretation 5:** Sales show distinct periodic surges around mid-month and month-end cycles, maintaining a healthy baseline demand throughout the period.
